```python
df["Server_Load"] = df["CPU_Utilization (%)"].apply(
    lambda x: "High Load" if x > 80 else "Normal Load"
)
```

```python
df.head()
```

```python
df["Log_Timestamp"] = pd.to_datetime(df["Log_Timestamp"], dayfirst=True)
```

```python
df["Log_Timestamp"].head()
```

```python
df["Year"] = df["Log_Timestamp"].dt.year
df["Month"] = df["Log_Timestamp"].dt.month
df["Day"] = df["Log_Timestamp"].dt.day
df["Hour"] = df["Log_Timestamp"].dt.hour
```

```python
df.to_csv("processed_server_data.csv", index=False)
```
